```python
import sys
from pathlib import Path
import torch

ROOT = Path.cwd().resolve()
while not (ROOT / 'modeling-common').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'modeling-common'))
import forecast_ablation as fa

ctx = fa.load_context(ROOT)
selection = fa.load_selection_contract(ROOT / 'configs' / 'gatgru' / 'feature_contract_v01.json', require_edge=True)
cfg = fa.load_baseline_config(ctx)
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
OUT = ROOT / 'artifacts' / 'gatgru' / 'v04_edge_controls'
summary = fa.run_fixed_variants(
    ctx, 'v04_edge_negative_controls', OUT,
    fa.edge_control_variants(cfg, selection['node_features'], selection['edge_features']),
    seeds=(0, 1, 2, 3, 4), device=DEVICE, score_weights=fa.BASELINE_SCORE_WEIGHTS,
    parent_version='v03_edge_ablation',
)
display(summary)
```
